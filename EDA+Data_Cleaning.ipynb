## Various Functions to plot different kinds of plot 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

%matplotlib inline
```

```python
################################# LINE PLOT  ####################################################

def line_plot(x, y, title="Line Plot", xlabel="X", ylabel="Y"):
    """
    Used to visualize trends over continuous data (time series, sequences).
    """
    plt.figure()
    plt.plot(x, y)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()

#################################################################################################  
    

################################# Scatter Plot  #################################################

def scatter_plot(x, y, title="Scatter Plot", xlabel="X", ylabel="Y"):
    """
    Used to visualize relationship or correlation between two variables.
    """
    plt.figure()
    plt.scatter(x, y)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()
    
################################################################################################# 


#################################  Bar Plot  ####################################################

def bar_plot(x, y, title="Bar Plot", xlabel="Categories", ylabel="Values"):
    """
    Used to compare quantities across categories.
    """
    plt.figure()
    plt.bar(x, y)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()

################################################################################################# 


#############################   Histogram Plot  #################################################

def histogram_plot(data, bins=10, title="Histogram", xlabel="Value", ylabel="Frequency"):
    """
    Used to visualize data distribution.
    """
    plt.figure()
    plt.hist(data, bins=bins)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()

################################################################################################# 


#################################  Box Plot  ####################################################

def box_plot(data, title="Box Plot", ylabel="Values"):
    """
    Used to detect outliers and spread of data.
    """
    plt.figure()
    plt.boxplot(data)
    plt.title(title)
    plt.ylabel(ylabel)
    plt.show()

################################################################################################# 


#################################  Pie Chart  ###################################################

def pie_chart(values, labels, title="Pie Chart"):
    """
    Used to show proportion of parts in a whole.
    """
    plt.figure()
    plt.pie(values, labels=labels, autopct="%1.1f%%")
    plt.title(title)
    plt.show()
    
################################################################################################# 


#################################   Area Plot  ##################################################

def area_plot(x, y, title="Area Plot", xlabel="X", ylabel="Y"):
    """
    Used to show cumulative magnitude over time.
    """
    plt.figure()
    plt.fill_between(x, y)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.show()

################################################################################################# 


############################## Multiple Line Plot  ##############################################

def multiple_line_plot(x, y_list, labels, title="Multiple Line Plot"):
    """
    Used to compare multiple trends on same axis.
    """
    plt.figure()
    for y, label in zip(y_list, labels):
        plt.plot(x, y, label=label)
    plt.legend()
    plt.title(title)
    plt.show()

################################################################################################# 


#################################  Line Plot  ###################################################

def sns_line_plot(df, x, y, title="Seaborn Line Plot"):
    """
    Better statistical line plot with confidence intervals.
    """
    sns.lineplot(data=df, x=x, y=y)
    plt.title(title)
    plt.show()
    
################################################################################################# 


################################# Scatter Plot  #################################################

def sns_scatter_plot(df, x, y, hue=None, title="Seaborn Scatter Plot"):
    """
    Scatter plot with grouping capability.
    """
    sns.scatterplot(data=df, x=x, y=y, hue=hue)
    plt.title(title)
    plt.show()

################################################################################################# 


#################################  Bar Plot  ####################################################

def sns_bar_plot(df, x, y, title="Seaborn Bar Plot"):
    """
    Aggregated bar plot (mean by default).
    """
    sns.barplot(data=df, x=x, y=y)
    plt.title(title)
    plt.show()
    
################################################################################################# 


#################################  Box Plot  ####################################################

def sns_box_plot(df, x, y, title="Seaborn Box Plot"):
    """
    Box plot with categorical separation.
    """
    sns.boxplot(data=df, x=x, y=y)
    plt.title(title)
    plt.show()
    
################################################################################################# 


#################################  Violin Plot  #################################################

def sns_violin_plot(df, x, y, title="Seaborn Violin Plot"):
    """
    Shows distribution shape + box plot info.
    """
    sns.violinplot(data=df, x=x, y=y)
    plt.title(title)
    plt.show()
    
################################################################################################# 


#################################  Hist Plot  ###################################################

def sns_hist_plot(df, x, bins=10, title="Seaborn Histogram"):
    """
    Distribution plot with KDE.
    """
    sns.histplot(data=df, x=x, bins=bins, kde=True)
    plt.title(title)
    plt.show()

################################################################################################# 


#################################  Heatmap  #####################################################

def sns_heatmap(data, title="Heatmap"):
    """
    Used to visualize matrix-like data (correlation, confusion matrix).
    """
    sns.heatmap(data, annot=True, fmt=".2f")
    plt.title(title)
    plt.show()
    
################################################################################################# 
    

################################# Pair Plot  ####################################################

def sns_pair_plot(df, hue=None):
    """
    Pairwise relationship visualization for entire dataset.
    """
    sns.pairplot(df, hue=hue)
    plt.show()
    
################################################################################################# 


################################# Count Plot  ###################################################

def sns_count_plot(df, x, title="Count Plot"):
    """
    Frequency of categorical variables.
    """
    sns.countplot(data=df, x=x)
    plt.title(title)
    plt.show()

################################################################################################# 
```

## Handling missing values 

Missing values should be fitted and transformed only on training data and should be fitted on testing data.

```python
def handle_missing_values(
    df,
    strategy="auto",                     # "auto", "drop", "fill" 
    numeric_strategy="mean",             # "mean", "median", "mode", "zero"
    categorical_strategy="mode",         # "mode", "constant"
    fill_value=None,
    drop_threshold=None,
    columns=None,
    verbose=True
):
    """
    Handles missing values in a DataFrame using multiple strategies.

    Parameters
    ----------
    df : pd.DataFrame
        Input DataFrame.

    strategy : str
        "auto"  -> numeric + categorical handled separately
        "drop"  -> drop rows or columns with missing values
        "fill"  -> fill with a single value (fill_value)

    numeric_strategy : str
        Strategy for numeric columns:
        "mean", "median", "mode", "zero"

    categorical_strategy : str
        Strategy for categorical columns:
        "mode", "constant"

    fill_value : any
        Used when strategy="fill" or categorical_strategy="constant"

    drop_threshold : float (0–1)
        Drop columns having missing ratio greater than threshold
        Example: 0.4 drops columns with >40% missing

    columns : list
        Columns to apply operation on (None → all)

    verbose : bool
        Print summary of operations

    Returns
    -------
    pd.DataFrame
        Cleaned DataFrame
    """

    df = df.copy()

    if columns is None:
        columns = df.columns

    # Drop columns based on missing percentage
    if drop_threshold is not None:
        missing_ratio = df.isna().mean()
        drop_cols = missing_ratio[missing_ratio > drop_threshold].index
        df.drop(columns=drop_cols, inplace=True)
        if verbose:
            print(f"Dropped columns due to missing threshold: {list(drop_cols)}")

    # Drop strategy
    if strategy == "drop":
        before = df.shape[0]
        df.dropna(subset=columns, inplace=True)
        after = df.shape[0]
        if verbose:
            print(f"Dropped {before - after} rows with missing values")

    # Fill with single value
    elif strategy == "fill":
        df[columns] = df[columns].fillna(fill_value)
        if verbose:
            print(f"Filled missing values with: {fill_value}")

    # Automatic handling
    elif strategy == "auto":
        for col in columns:
            if df[col].isna().sum() == 0:
                continue

            # Numeric columns
            if pd.api.types.is_numeric_dtype(df[col]):
                if numeric_strategy == "mean":
                    value = df[col].mean()
                elif numeric_strategy == "median":
                    value = df[col].median()
                elif numeric_strategy == "mode":
                    value = df[col].mode()[0]
                elif numeric_strategy == "zero":
                    value = 0
                df[col].fillna(value, inplace=True)

                if verbose:
                    print(f"{col}: filled numeric missing with {numeric_strategy} ({value})")

            # Categorical columns
            else:
                if categorical_strategy == "mode":
                    value = df[col].mode()[0]
                elif categorical_strategy == "constant":
                    value = fill_value if fill_value is not None else "Unknown"
                df[col].fillna(value, inplace=True)

                if verbose:
                    print(f"{col}: filled categorical missing with {value}")

    return df
```

## Handling duplicates

Always handle duplicates before train-test split

```python
def handle_duplicates(
    df,
    subset=None,
    keep="first",       # 'first', 'last', or False
    strategy="drop",    # 'drop', 'aggregate'
    agg_dict=None,
    verbose=True
):
    """
    Handles duplicate rows in a DataFrame.

    Parameters
    ----------
    df : pd.DataFrame
        Input DataFrame

    subset : list or None
        Columns to consider for duplicate detection
        None -> considers all columns

    keep : str
        'first', 'last', or False
        Determines which duplicate to keep

    strategy : str
        'drop' -> drop duplicate rows
        'aggregate' -> aggregate duplicates using agg_dict

    agg_dict : dict
        Required when strategy='aggregate'
        Example:
        {'Salary': 'mean', 'Age': 'max'}

    verbose : bool
        Print summary of operation

    Returns
    -------
    pd.DataFrame
        DataFrame after duplicate handling
    """

    df = df.copy()

    # Count duplicates
    dup_count = df.duplicated(subset=subset).sum()

    if verbose:
        print(f"Duplicate rows detected: {dup_count}")

    if dup_count == 0:
        return df

    # Strategy 1: Drop duplicates
    if strategy == "drop":
        df = df.drop_duplicates(subset=subset, keep=keep)
        if verbose:
            print(f"Dropped duplicates, kept='{keep}'")

    # Strategy 2: Aggregate duplicates
    elif strategy == "aggregate":
        if agg_dict is None:
            raise ValueError("agg_dict must be provided for aggregation")

        if subset is None:
            raise ValueError("subset must be provided for aggregation")

        df = (
            df.groupby(subset, as_index=False)
              .agg(agg_dict)
        )

        if verbose:
            print("Aggregated duplicate rows using provided agg_dict")

    return df
```

## Handling Imbalenced Dataset


I handle imbalanced datasets by resampling only the training data using oversampling, downsampling, or SMOTE depending on dataset size. This prevents data leakage and ensures the test set represents real-world distribution.

```python
from collections import Counter
from sklearn.utils import resample
from imblearn.over_sampling import SMOTE

def handle_imbalanced_data(
    X,
    y,
    strategy="auto",
    sampling_strategy="minority",
    random_state=42,
    verbose=True
):
    """
    Handles class imbalance in a dataset.

    Parameters
    ----------
    X : pd.DataFrame or np.ndarray
        Feature matrix (TRAIN DATA ONLY)

    y : pd.Series or np.ndarray
        Target variable

    strategy : str
        'undersample' -> reduce majority class           # Downsampling
        'oversample'  -> duplicate minority class        # Upsampling
        'smote'       -> synthetic sample generation
        'auto'        -> chooses oversample for small data, smote for large

    sampling_strategy : str or float
        Used by SMOTE / resample
        'minority', 'auto', or ratio

    random_state : int
        Ensures reproducibility

    verbose : bool
        Print class distribution before & after

    Returns
    -------
    X_resampled, y_resampled
    """

    if verbose:
        print("Before balancing:", Counter(y))

    # AUTO STRATEGY
    if strategy == "auto":
        if len(y) < 10000:
            strategy = "oversample"
        else:
            strategy = "smote"

    # UNDERSAMPLING
    if strategy == "undersample":
        df = X.copy()
        df["target"] = y

        majority_class = df["target"].value_counts().idxmax()
        minority_class = df["target"].value_counts().idxmin()

        df_majority = df[df["target"] == majority_class]
        df_minority = df[df["target"] == minority_class]

        df_majority_downsampled = resample(
            df_majority,
            replace=False,
            n_samples=len(df_minority),
            random_state=random_state
        )

        df_balanced = pd.concat([df_majority_downsampled, df_minority])

        X_resampled = df_balanced.drop("target", axis=1)
        y_resampled = df_balanced["target"]

    # OVERSAMPLING
    elif strategy == "oversample":
        df = X.copy()
        df["target"] = y

        majority_class = df["target"].value_counts().idxmax()
        minority_class = df["target"].value_counts().idxmin()

        df_majority = df[df["target"] == majority_class]
        df_minority = df[df["target"] == minority_class]

        df_minority_upsampled = resample(
            df_minority,
            replace=True,
            n_samples=len(df_majority),
            random_state=random_state
        )

        df_balanced = pd.concat([df_majority, df_minority_upsampled])

        X_resampled = df_balanced.drop("target", axis=1)
        y_resampled = df_balanced["target"]

    # SMOTE
    elif strategy == "smote":
        smote = SMOTE(
            sampling_strategy=sampling_strategy,
            random_state=random_state
        )
        X_resampled, y_resampled = smote.fit_resample(X, y)

    else:
        raise ValueError("Invalid strategy selected")

    if verbose:
        print("After balancing:", Counter(y_resampled))

    return X_resampled, y_resampled
```

## Handling of Data Types

Always convert data types before:

-> Scaling

-> Encoding

-> Model training

-> SMOTE

I convert data types explicitly where required and infer numeric, datetime, and categorical types automatically to prevent silent errors and optimize memory before modeling.

```python
def convert_dtypes(
    df,
    dtype_map=None,
    infer_numeric=True,
    parse_datetime=True,
    categorical_threshold=0.05,
    errors="coerce",
    verbose=True
):
    """
    Converts column data types in a DataFrame safely.

    Parameters
    ----------
    df : pd.DataFrame
        Input DataFrame

    dtype_map : dict or None
        Explicit dtype mapping
        Example:
        {"age": "int", "salary": "float", "gender": "category"}

    infer_numeric : bool
        Try converting object columns to numeric

    parse_datetime : bool
        Try parsing datetime columns automatically

    categorical_threshold : float
        If unique_ratio < threshold, convert to category

    errors : str
        'raise', 'ignore', 'coerce'

    verbose : bool
        Print conversion summary

    Returns
    -------
    pd.DataFrame
        DataFrame with optimized dtypes
    """

    df = df.copy()

    # 1️⃣ Explicit conversions
    if dtype_map:
        for col, dtype in dtype_map.items():
            if col in df.columns:
                try:
                    df[col] = df[col].astype(dtype)
                    if verbose:
                        print(f"{col}: converted to {dtype}")
                except Exception as e:
                    if verbose:
                        print(f"{col}: conversion failed ({e})")

    # 2️⃣ Datetime inference
    if parse_datetime:
        for col in df.select_dtypes(include="object"):
            try:
                converted = pd.to_datetime(df[col], errors="coerce")
                if converted.notna().sum() > 0:
                    df[col] = converted
                    if verbose:
                        print(f"{col}: parsed as datetime")
            except Exception:
                pass

    # 3️⃣ Numeric inference
    if infer_numeric:
        for col in df.select_dtypes(include="object"):
            converted = pd.to_numeric(df[col], errors=errors)
            if converted.notna().sum() > 0:
                df[col] = converted
                if verbose:
                    print(f"{col}: converted to numeric")

    # 4️⃣ Convert low-cardinality columns to category
    for col in df.select_dtypes(include=["int64", "object"]):
        unique_ratio = df[col].nunique() / len(df)
        if unique_ratio < categorical_threshold:
            df[col] = df[col].astype("category")
            if verbose:
                print(f"{col}: converted to category")

    return df
```

## Data Encoding 

Categorical variables are encoded using one-hot or ordinal encoding depending on semantics. Encoders are fitted on training data only and reused for test data to prevent leakage.

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder

###################### For train data #######################################

def fit_encoders(
    X_train,
    y_train=None,
    onehot_cols=None,
    ordinal_cols=None,
    ordinal_order=None,
    label_cols=None,
    target_cols=None,
    smoothing=1.0
):
    """
    Learns encoding parameters from TRAIN data only.

    Returns a dictionary of fitted encoders / mappings.
    """

    encoders = {}

    # ---------- One-Hot ----------
    if onehot_cols:
        ohe = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False,
            drop="first"   # avoids dummy variable trap
        )
        ohe.fit(X_train[onehot_cols])
        encoders["onehot"] = {"cols": onehot_cols, "encoder": ohe}

    # ---------- Ordinal ----------
    if ordinal_cols:
        if ordinal_order is None:
            raise ValueError("Provide ordinal_order for ordinal encoding")

        oe = OrdinalEncoder(categories=ordinal_order)
        oe.fit(X_train[ordinal_cols])
        encoders["ordinal"] = {"cols": ordinal_cols, "encoder": oe}

    # ---------- Label Encoding ----------
    if label_cols:
        label_encoders = {}
        for col in label_cols:
            le = LabelEncoder()
            le.fit(X_train[col].astype(str))
            label_encoders[col] = le

        encoders["label"] = {"cols": label_cols, "encoders": label_encoders}

    # ---------- Target / Mean ---------------
    if target_cols:
        if y_train is None:
            raise ValueError("y_train required for target encoding")

        target_maps = {}
        global_mean = y_train.mean()

        for col in target_cols:
            stats = (
                pd.concat([X_train[col], y_train], axis=1)
                .groupby(col)[y_train.name]
                .mean()
            )
            # smoothing to reduce overfitting
            target_maps[col] = {
                "map": (stats * smoothing + global_mean) / (smoothing + 1),
                "global_mean": global_mean
            }

        encoders["target"] = {"cols": target_cols, "maps": target_maps}

    return encoders

#############################################################################


###################### For train and test data ##############################

def transform_encoders(X, encoders):
    """
    Applies learned encoders to a dataset (TRAIN or TEST).
    """

    X = X.copy()

    # ---------- One-Hot ----------
    if "onehot" in encoders:
        cols = encoders["onehot"]["cols"]
        ohe = encoders["onehot"]["encoder"]

        ohe_df = pd.DataFrame(
            ohe.transform(X[cols]),
            columns=ohe.get_feature_names_out(cols),
            index=X.index
        )

        X.drop(columns=cols, inplace=True)
        X = pd.concat([X, ohe_df], axis=1)

    # ---------- Ordinal ----------
    if "ordinal" in encoders:
        cols = encoders["ordinal"]["cols"]
        oe = encoders["ordinal"]["encoder"]
        X[cols] = oe.transform(X[cols])

    # ---------- Label ----------
    if "label" in encoders:
        for col, le in encoders["label"]["encoders"].items():
            # unseen labels handled safely
            X[col] = X[col].astype(str)
            X[col] = X[col].map(
                lambda x: le.transform([x])[0] if x in le.classes_ else -1
            )

    # ---------- Target ----------
    if "target" in encoders:
        for col, info in encoders["target"]["maps"].items():
            X[col] = X[col].map(info["map"]).fillna(info["global_mean"])

    return X

#############################################################################
```

## Handling Outliers

Outliers are handled by learning thresholds from training data using IQR or quantile methods, then applying capping or removal consistently to both training and test sets to prevent data leakage.

```python

###################### For train data #######################################

def fit_outlier_handler(
    X_train,
    cols=None,
    method="iqr",          # "iqr", "zscore", "quantile" 
    factor=1.5,
    lower_quantile=0.01,
    upper_quantile=0.99
):
    """
    Learns outlier thresholds from TRAIN data only.

    method: "iqr", "zscore", "quantile" 
    
    Returns a dictionary of thresholds.
    """

    thresholds = {}

    if cols is None:
        cols = X_train.select_dtypes(include=["int64", "float64"]).columns

    for col in cols:
        series = X_train[col].dropna()

        if method == "iqr":
            Q1 = series.quantile(0.25)
            Q3 = series.quantile(0.75)
            IQR = Q3 - Q1
            lower = Q1 - factor * IQR
            upper = Q3 + factor * IQR

        elif method == "zscore":
            mean = series.mean()
            std = series.std()
            lower = mean - factor * std
            upper = mean + factor * std

        elif method == "quantile":
            lower = series.quantile(lower_quantile)
            upper = series.quantile(upper_quantile)

        else:
            raise ValueError("Invalid outlier detection method")

        thresholds[col] = (lower, upper)

    return thresholds

#############################################################################


###################### For train and test data ##############################

def transform_outliers(
    X,
    thresholds,
    strategy="cap"
):
    """
    Applies learned outlier thresholds.

    strategy:
    - 'cap'  -> clip values
    - 'remove' -> remove rows with outliers
    """

    X = X.copy()

    for col, (lower, upper) in thresholds.items():
        if col not in X.columns:
            continue

        if strategy == "cap":
            X[col] = X[col].clip(lower, upper)

        elif strategy == "remove":
            X = X[(X[col] >= lower) & (X[col] <= upper)]

        else:
            raise ValueError("Invalid strategy")

    return X

#############################################################################
```
